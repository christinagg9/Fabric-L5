```python
import pandas as pd

# Create a basic dataset
data = {
    'employee_id': range(1, 21),
    'name': ['John', 'Jane', 'Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ivy', 'Jack', 'Karen', 'Leo', 'Mia', 'Nina', 'Olivia', 'Paul', 'Quinn', 'Rita'],
    'age': [22, 25, 28, 35, 40, 42, 50, 60, 65, 30, 28, 24, 37, 41, 48, 33, 29, 32, 45, 38],
    'department': ['HR', 'Engineering', 'Marketing', 'Finance', 'Sales', 'HR', 'Engineering', 'Marketing', 'Finance', 'Sales', 'HR', 'Engineering', 'Marketing', 'Finance', 'Sales', 'HR', 'Engineering', 'Marketing', 'Finance', 'Sales'],
    'salary': [50000, 60000, 55000, 75000, 80000, 85000, 100000, 110000, 120000, 65000, 67000, 72000, 85000, 90000, 95000, 48000, 67000, 70000, 80000, 88000],
    'work_experience': [1, 3, 5, 7, 10, 12, 15, 18, 20, 3, 1, 6, 4, 8, 5, 7, 3, 2, 6, 5],
}

# Convert the basic dataset to a Dataframe
df = pd.DataFrame(data)

# Display the first few rows of the dataset
display(df.head(3))

```

```python
# Check for missing values 
# Using isnull() alias:

missing_values = df.isnull()

display(missing_values) # Display false = non-null cells
```

```python
# Summary statistics of numeric columns (age, salary, work_experience)

# Age
mean_age = df['age'].mean() # Gives the average of the age column
median_age = df['age'].median() # Gives the middle value when sorted
mode_age = df['age'].mode() # Gives the most frequent value 
std_age = df['age'].std() # Measures how spread out the values are from the average

# Salary
mean_salary = df['salary'].mean() # Gives the average of the salary column
median_salary = df['salary'].median() # Gives the middle value when sorted
mode_salary = df['salary'].mode() # Gives the most frequent value
std_salary = df['salary'].std() # Measures how spread out the values are from the average

#Work_Experience
mean_work_experience = df['work_experience'].mean() # Gives the average of the work_experience column
median_work_experience = df['work_experience'].median() # Gives the middle value when sorted
mode_work_experience = df['work_experience'].mode() # Gives the most frequent value
std_work_experience = df['work_experience'].std() # Measures how spread out the values are from the average

# Display results
print(f"Age = Mean Age: {mean_age}, Median Age: {median_age}, Mode Age: {mode_age}, Std Age: {std_age}")
print(f"Salary = Mean Salary: {mean_salary}, Median Salary: {median_salary}, Mode Salary: {mode_salary}, Std Salary: {std_salary}")
print(f"Work_Experience = Mean Work_Experience: {mean_work_experience}, Median Work_Experience: {median_work_experience}, Mode Work_Experience: {mode_work_experience}, Std Work_Experience: {std_work_experience}")
```

### Task 3: Data Transformation

Objective: Perform basic data transformations such as creating new columns and renaming existing ones.

Instructions:

Create a new column called years_to_retirement by subtracting the age from 65.

Rename the salary column to annual_salary.

Sort the DataFrame based on work_experience in descending order.

```python
# Create a new column called years_to_retirement by subtracting the age from 65

df['years_to_retirement'] = df['age'].apply(lambda x: 65 - x)
display(df)
```

```python
# Rename the salary column to annual_salary

df.rename(columns={'salary' : 'annual_salary'}, inplace=True) # inplace=True permanantly renames column in df
display(df)
```

```python
# Sort the DataFrame based on work_experience in descending order

sorted_df = df.sort_values(by='work_experience', ascending=False) #ascending=False: descending order (Highest to lowest)
display(sorted_df)
```

### Task 4: Handling Missing Data

Objective: Handle missing data (if any).

Instructions:

Introduce some missing values in the annual_salary column.

Handle the missing data by either filling it with the mean salary or dropping the rows with missing values.

```python
# Introduce missing values in the annual_salary column
# Using replace() to introduce missing values to existing annual salary values with NULL/None 

df['annual_salary'] = df['annual_salary'].replace([50000,65000,80000,95000], None)

display(df)
```

```python
# Handle the missing data by filling it with the mean salary 
# Using fillna()

#df['annual_salary'] = df['annual_salary'].fillna(df['annual_salary'].mean())
#display(df)

# Not executed. 
```

```python
# Handle the missing data by dropping the rows with missing values
# Using dropna ()

df_cleaned = df.dropna()

# Display cleaned dataframe
display(df_cleaned)
```

## Task 5: Data Filtering and Selection
Objective: Select specific data based on conditions.

Instructions:

Filter the data to show only employees who are older than 30.

Select only the columns name, age, and annual_salary.

```python
# Filter data to show only employees who are older than 30

filtered_df = df_cleaned[df_cleaned['age'] > 30]

# Select only the columns name, age and annual_salary

display(filtered_df[['name','age','annual_salary']])
```

### Task 6: Data Aggregation
Objective: Perform group-based aggregation.

Instructions:

Group the data by department and calculate the average salary for each department.

```python
# Group the data by department and calculate the average salary for each department
# GroupBy: Average salary for each department

# .mean() calculates the average salary for each department
# .reset_index() displays 'department' as a regular table column
average_salary_by_department = df_cleaned.groupby('department')[['annual_salary']].mean().reset_index() 

# rename column 'annual_salary' to 'average_salary' for cleaner result output purposes  
display(average_salary_by_department.rename(columns={'annual_salary' : 'average_salary'}))
```

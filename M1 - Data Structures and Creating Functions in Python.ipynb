```python
# Create two sets and perform a union and an intersection operation between them
#Set Operations
my_set1 = {3, 6, 9, 15}
my_set2 = {5, 15, 20}

#Set Union (Combine Sets)
my_set1 = {3, 6, 9, 15}
my_set2 = {5, 15, 20}
print(my_set1 | my_set2) # Output: {3, 20, 5, 6, 9, 15} # Unordered

# Set Intersection (Common Element)
print(my_set1 & my_set2) #Output: {15}
```
